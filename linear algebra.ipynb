```python
import sympy as sp 

u=sp.symbols('u')
t=sp.symbols('t')
lam=sp.symbols('lambda')
#The constant of integration
C=sp.symbols('C')

#Left hand side integration
lhs_integrated=sp.integrate(1/u,u)
print(f"Integrated LHS (Left side):{lhs_integrated}")    

#Right hand side integration
rhs_integrated=sp.integrate(lam,t)
print(f"Integrated RHS (Right side):{rhs_integrated}")

integrated_eq=sp.Eq(lhs_integrated,rhs_integrated+C)

print("\n--- 2. The Integrated Result---")
sp.pprint(integrated_eq)

u_solved=sp.solve(integrated_eq,u)

print("\n--- 3. Solving for u (The Final Exponential) ---") 

sp.pprint(sp.Eq(u, u_solved[0]))     
```

```python
import numpy as np

# Step 1: Data (2D example)
X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0]
])

# Step 2: Mean center
mean = np.mean(X, axis=0)
X_centered = X - mean

# Step 3: Covariance matrix
cov_matrix = np.cov(X_centered.T)

# Step 4: Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Step 5: Sort by importance
idx = np.argsort(eigenvalues)[::-1]
eigenvectors = eigenvectors[:, idx]

# Step 6: Project data (reduce to 1D)
X_reduced = X_centered.dot(eigenvectors[:, 0])

print("Reduced Data:")
print(X_reduced)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. The Physical "Machine" Parameters ---
mass = 1.0       # kg
k_spring = 2.0   # Spring stiffness (Newton/meter)
c_damper = 0.5   # Damping friction (Newton*s/meter)

# --- 2. The Linear Algebra (State-Space Form) ---
# We convert physics into a Matrix 'A'
# System State x = [position, velocity]
# dx/dt = A * x
#
# Matrix A structure for Mass-Spring-Damper:
# [  0      1   ]  -> change in position is velocity
# [ -k/m  -c/m  ]  -> change in velocity is Force/mass (Newton's 2nd Law)

A = np.array([
    [0, 1],
    [-k_spring/mass, -c_damper/mass]
])

print("System Matrix A:")
print(A)

# --- 3. The "Stability Check" (Eigenvalues) ---
# If the real part of eigenvalues is negative, the system is stable.
# If positive, the machine explodes/crashes.
eigenvals, eigenvecs = np.linalg.eig(A)
print("\nEigenvalues of the machine:")
print(eigenvals)

# --- 4. The Simulation Loop (Time Integration) ---
dt = 0.01  # Time step (seconds)
T_max = 20 # Total time to run
steps = int(T_max / dt)

# Initial State: [Position = 5 meters, Velocity = 0 m/s]
# We are pulling the mass back and letting go.
state = np.array([5.0, 0.0])

# Log data for plotting
positions = []
times = []

# Identity Matrix (needed for discrete time update)
I = np.eye(2)

print("\nSimulating machine physics...")
for t in range(steps):
    # Log current position
    positions.append(state[0])
    times.append(t * dt)
    
    # UPDATE STEP: The "Engine" of the simulation
    # state_new = state_old + (Change * dt)
    # dx/dt = A * x
    # So: x_new = x + (A * x) * dt
    # This can be rewritten using matrix algebra as: x_new = (I + A*dt) * x
    
    # This is "Euler Integration" - the simplest form of calculus in code
    state = state + (A @ state) * dt

# --- 5. Visualization ---
plt.figure(figsize=(10, 6))
plt.plot(times, positions, label='Position (meters)', color='blue', linewidth=2)
plt.axhline(0, color='black', linewidth=1, linestyle='--')
plt.title(f"Machine Response (Mass-Spring-Damper)\nEigenvalues: {eigenvals[0]:.2f}, {eigenvals[1]:.2f}")
plt.xlabel("Time (seconds)")
plt.ylabel("Displacement")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import sympy as sp 
from sympy import Matrix,eye,factorial
import scipy.linalg
A=Matrix([
    [0,-1],
    [1,0],
])
x=sp.symbols('x')
A_np=np.array(A).astype(float)
solution=scipy.linalg.expm(A_np)

print("Exponential of matrix A:")
print(solution)

def matrix_exponential(A, n_terms=1000):
    A=Matrix(A)
    I=eye(A.rows)
    result=I
    for n in range(1, n_terms):
        result += (A**n) / factorial(n)
    return result
print("Matrix exponential of A using Taylor series:")
print(matrix_exponential(A.evalf(4)))
```

```python
x=sp.symbols('x')
x_looped = x % (2 * sp.pi)
func = sp.Piecewise(
    (1, x_looped < sp.pi), 
    (-1, True))

n_terms=np.arange(100)
fs_eq=0
z=0
u=2*sp.pi



#Calculate period
def calculate_period(func,i=0,max=4*np.pi,dx=0.01):
    
    if i>=max:
        return None
    epsilon_y = 0.01
    #Define starting points
    n_start=np.arange(3)
    starting_points=[]
    
    for n in n_start:
        starting_points.append(i+(n*0.5))
        
    
    x_intervals=[]
    #Define X
    for idx, j in enumerate(starting_points):
    # Handle the last anchor item
        if idx == len(starting_points) - 1:
            x_intervals.append(np.arange(j, j + max, dx))
        else:
            # Safely look up the next item using the clean integer index
            x_intervals.append(np.arange(j,j+max, dx))

    periods=[]
    #Iterate looking for y 

    for a in x_intervals:
        for x_val in a:
            #check for the first value
            y_now=func.subs({x:x_val}).evalf()
            y_next=func.subs({x:x_val+0.001}).evalf()
            diff=(y_next-y_now)/0.001
            if x_val in starting_points:
                x_start=x_val
                y_start=func.subs({x:x_val}).evalf()
                diff_start=diff
                continue
            if x_val-x_start < 0.2:
                continue
            y_value=func.subs({x:x_val}).evalf()
        #for each y find derivative
            diff_value=diff
        #Then check for similarity in value and directio
            if abs(y_start-y_value)<epsilon_y and ((diff_start*diff_value)>0 or(diff_start==0 and diff_value==0 )):
                periods.append(round(x_val-x_start,4))
                break
    #If we didn't find the period for each starting point redo until i reaches max        
    if len(periods)<len(starting_points):
        return calculate_period(func,i+0.5)
    #Also redo if values of period of each starting point are not equal    
    first_period=periods[0]                
    for period in periods:
        if first_period!=period:
                return calculate_period(func,i+0.5)    
    return periods[0]

# --- Dynamic Dual-Check Intervalling ---
L = sp.periodicity(func, x)
if L is None:
    print("SymPy failed. Triggering custom numerical scanner fallback...")
    L = calculate_period(func)

# FIX: Add a checkpoint before the Fourier Loop starts!
if L is None:
    print("\n[System Notice]: The function is completely APERIODIC.")
    print("A standard Fourier Series cannot be calculated because no period exists.")
    print("Switching strategy: To analyze a non-repeating signal, you need a Fourier TRANSFORM.")
    # Exit the script cleanly instead of running into line 73 and crashing
    import sys
    sys.exit()

# If L is a real number, the script continues down to your Fourier loop smoothly...

#N terms i.e cos(nx)
for n in n_terms:
    #Handle edge case n=0
    if n==0:
        a0=sp.integrate(func,(x,0,L))
        fs_eq+=(a0/L)
    #Evaluate  for sine
    b=sp.integrate(func*sp.sin((n*2*sp.pi*x)/L),(x,0,L))
    fs_eq+=((b*sp.sin((n*2*sp.pi*x)/L))/(L/2))
    
    #Evaluate for cosine
    a=sp.integrate(func*sp.cos((n*2*sp.pi*x)/L),(x,0,L))
    fs_eq+=((a*sp.cos((n*2*sp.pi*x)/L))/(L/2))
    
original_eq=func.evalf()
final_eq=fs_eq.evalf()

original_numpy=sp.lambdify(x,original_eq,'numpy')
final_numpy=sp.lambdify(x,final_eq,'numpy')
x_vals=np.linspace(-3*np.pi,3*np.pi,500)
y_original=original_numpy(x_vals)
y_final=final_numpy(x_vals)

latex_original=sp.latex(func)
latex_final=sp.latex(fs_eq)
plt.plot(x_vals,y_original,label=rf"Original eqn ", linestyle="--", color="blue", linewidth="4")
plt.plot(x_vals,y_final,label=rf"Final equation " ,color="orange" ,linewidth="2")
plt.title("Fourier series")
plt.legend()
plt.show()

```

```python
original_size=img_gray.nbytes
U,E,V_T = np.linalg.svd(img_gray,full_matrices=False)
E=np.diag(E)
plt.figure(figsize=(15,15))
ranks=[1,10,20,50,100,200]
for i,r in enumerate(ranks):
    U_r=U[:,:r]
    E_r=E[:r,:r]
    V_T_r=V_T[:r,:]
    compressed_size=U_r.nbytes+E_r.nbytes+V_T_r.nbytes
    compression_percentage=compressed_size/original_size*100
    img_reconstructed=U_r@E_r@V_T_r
    plt.subplot(3,2,i+1)
    plt.title(f"Rank {r} Image size: {compression_percentage:.2f}%")
    plt.axis('off')
    plt.imshow(img_reconstructed,cmap='gray')
plt.show()
```

```python
# Strip away the zeros and get the flat 1D array of singular values
singular_values_array = np.diag(E)
#Get an array of indexes of the singular values
singular_value_indices = np.arange(len(singular_values_array))

# Plot the singular values
plt.figure(figsize=(10, 6))
plt.plot(singular_value_indices, singular_values_array, marker='o', linestyle='-', color='blue')
plt.title("Singular Values of the Image")
plt.xlabel("Index of Singular Value")
plt.ylabel("Singular Value Magnitude")
plt.show()
```

```python
import random 
colours=['red','green','blue','orange','purple','cyan','magenta']
#find log of singular values
log_singular_values = np.log(singular_values_array)

# Plot the log of singular values
plt.figure(figsize=(10, 6))
plt.plot(singular_value_indices, log_singular_values, marker='o', linestyle='-', color='orange')
#plot a vertical for every rank
for r in ranks:
    random_color = random.choice(colours)
    plt.axvline(r, color=random_color, linestyle='--', label=f'Rank {r}')

plt.title("Log of Singular Values of the Image")
plt.xlabel("Index of Singular Value")
plt.ylabel("Log of Singular Value Magnitude")
plt.legend()
plt.show()
```

```python
# add cumulative singular values at every index
total=np.sum(singular_values_array)
cumulative_singular_values = np.cumsum(singular_values_array)/total
plt.plot(singular_value_indices, cumulative_singular_values, marker='o', linestyle='-', color='green')
for r in ranks:
    random_color = random.choice(colours)
    plt.axvline(r, color=random_color, linestyle='--', label=f'Rank {r}')
    #print the value
    print(f"Percentage of information at rank {r} is: {100*cumulative_singular_values[r]:.4f} %")
plt.title("Fraction of Cumulative Singular Values of the Image")
plt.xlabel("Index of Singular Value")
plt.legend()
plt.ylabel("Cumulative Sum of Singular Values")
plt.show()
```
