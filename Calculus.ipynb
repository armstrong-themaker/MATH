```python
#Find limits 
import sympy as sp 
from sympy.plotting import plot
x = sp.Symbol('x')
expr=sp.sin(5*x)/x
limit=sp.limit(expr,x,0)
print(f"Limit:{limit}")

plot(expr, (x, -1, 1), title="Limit approaching 5")
```

```python
import sympy as sp 
x=sp.Symbol('x')
expr=sp.sin(sp.sqrt(x+sp.sqrt(3*x)))
diff=sp.diff(expr,x)
print(f"Derivative: {diff}")
```

```python
import sympy as sp 
import numpy as np 
import matplotlib.pyplot as plt
x=sp.symbols('x')
expr=2*x**3+12*x**2+20*x+56
diff1=sp.diff(expr,x,1)
diff2=sp.diff(expr,x,2)

#convert to numpy function
diff1_func=sp.lambdify(x,diff1,'numpy')
diff2_func=sp.lambdify(x,diff2,'numpy')

#generate x and y values
x_vals=np.linspace(-10,10,100)
diff1_y=diff1_func(x_vals)
diff2_y=diff2_func(x_vals)
#plot
plt.subplot(2,1,1)
plt.plot(x_vals,diff1_y)
plt.title("DERIVATIVE GRAPH")
plt.xlabel('x')
plt.ylabel('derivative')

plt.subplot(2,1,2)
plt.plot(x_vals,diff2_y)
plt.title("SECOND DERIVATIVE GRAPH")
plt.xlabel('x')
plt.ylabel('Second derivative')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# x values for the right branch
x_right = np.linspace(1, 5, 500)

# x values for the left branch
x_left = np.linspace(-5, -1, 500)

# Hyperbola: x² - y² = 1
y_right = np.sqrt(x_right**2 - 1)
y_left = np.sqrt(x_left**2 - 1)

plt.figure(figsize=(8,8))

# Upper branches
plt.plot(x_right, y_right, label=r"$x^2-y^2=1$")
plt.plot(x_left, y_left)

# Lower branches
plt.plot(x_right, -y_right)
plt.plot(x_left, -y_left)

# Coordinate axes
plt.axhline(0)
plt.axvline(0)

plt.xlim(-5,5)
plt.ylim(-5,5)

plt.gca().set_aspect('equal')

plt.grid(True)
plt.title(r"Hyperbola: $x^2-y^2=1$")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-3, 3, 500)

x = (np.exp(t) + np.exp(-t)) / 2  # x = cosh(t)
y = (np.exp(t) - np.exp(-t)) / 2  # y = sinh(t)

plt.figure(figsize=(8,8))

# Right branch
plt.plot(x, y, label="x = cosh(t), y = sinh(t)")

# Left branch (mirror image)
plt.plot(-x, y)
plt.plot(t,t,label="y=x")
plt.axhline(0)
plt.axvline(0)

plt.gca().set_aspect('equal')
plt.grid(True)

plt.title("Hyperbola Generated Using Hyperbolic Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python
x=sp.symbols('x')
eq=(3*x+2)/(x-5)
numpy_eq=sp.lambdify(x,eq,'numpy')
x=np.linspace(-10,10,100)
y=numpy_eq(x)

plt.plot(x,y,label="y=(2x+2)/(x-5)")
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.axhline(0)
plt.axvline(0)

plt.grid(True)

plt.title("Rational Function Graph")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python
x=sp.symbols('x')
expr=(sp.exp(x)-sp.exp(-x))
derivative=sp.diff(expr, x)
print(f"Derivative: {derivative}")
```

```python
x=sp.symbols('x')
expr=sp.exp(x)*sp.cos(2*x)
integral=sp.integrate(expr, x)
print(f"Integral: {integral}")
```
